```python
import pandas as pd
import numpy as np
import os
import gensim
import nltk
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
folder_path = r"/Users/ruchisehgal/Documents/Study Material/Gen AI and Agentic AI/Embedding/data"
story = []
for filename in os.listdir(folder_path):
  file_path=os.path.join(folder_path,filename)
  with open(file_path,encoding='unicode_escape') as f:
    corpus=f.read()
  raw_sent=sent_tokenize(corpus)
  for sent in raw_sent:
    story.append(simple_preprocess(sent))
```

```python
story
```

```python
custom_model =gensim.models.Word2Vec(
    window=16,
    min_count=1,
    vector_size=210
    
)
#W and V are capital
```

```python
custom_model.build_vocab(story)
print(custom_model.corpus_count)
```

```python
custom_model.train(story, total_examples=custom_model.corpus_count, epochs=21)
#epoch--hyperparameter ---no. of times repeat learnings on data here 
# 40720--no. of docs
# 49980-- no.of unique words
```
